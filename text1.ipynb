```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, LabelEncoder, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import r2_score


df = pd.read_csv("beer-servings.csv")


df.dropna(inplace=True)


le = LabelEncoder()
for col in df.select_dtypes(include=['object']).columns:
    df[col] = le.fit_transform(df[col])


X = df.drop("total_litres_of_pure_alcohol", axis=1)
y = df["total_litres_of_pure_alcohol"]


scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)


poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X_scaled)


X_train, X_test, y_train, y_test = train_test_split(
    X_poly, y, test_size=0.25, random_state=42
)


models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=3.0),
    "Lasso Regression": Lasso(alpha=0.1)
}


results = []

for model_name, model in models.items():
    model.fit(X_train, y_train)

    r2_train = r2_score(y_train, model.predict(X_train))
    r2_test = r2_score(y_test, model.predict(X_test))

    results.append([model_name, "Train R2 Score", r2_train])
    results.append([model_name, "Test R2 Score", r2_test])


results_df = pd.DataFrame(
    results,
    columns=["Model Name", "Performance Metric", "Metric Value"]
)

results_df
```
