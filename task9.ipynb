```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Linear
def linear(x):
    return x

# ReLU
def relu(x):
    return np.maximum(0, x)

# Tanh
def tanh(x):
    return np.tanh(x)

# Softmax
def softmax(x):
    exp_x = np.exp(x - np.max(x))  # numerical stability
    return exp_x / np.sum(exp_x)
```

```python
x = np.linspace(-10, 10, 400)
```

```python
plt.plot(x, sigmoid(x))
plt.title("Sigmoid Activation Function")
plt.xlabel("x")
plt.ylabel("Sigmoid(x)")
plt.grid(True)
plt.show()
```

```python
plt.plot(x, linear(x))
plt.title("Linear Activation Function")
plt.xlabel("x")
plt.ylabel("Linear(x)")
plt.grid(True)
plt.show()
```

```python
plt.plot(x, relu(x))
plt.title("ReLU Activation Function")
plt.xlabel("x")
plt.ylabel("ReLU(x)")
plt.grid(True)
plt.show()
```

```python

plt.plot(x, tanh(x))
plt.title("Tanh Activation Function")
plt.xlabel("x")
plt.ylabel("Tanh(x)")
plt.grid(True)
plt.show()
```

```python
# Softmax is usually applied to vectors
x_softmax = np.linspace(-5, 5, 10)
y_softmax = softmax(x_softmax)

plt.bar(x_softmax, y_softmax)
plt.title("Softmax Activation Function")
plt.xlabel("x")
plt.ylabel("Softmax(x)")
plt.grid(True)
plt.show()
```
