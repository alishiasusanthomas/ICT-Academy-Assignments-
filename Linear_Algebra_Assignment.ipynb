### Question 1
Using auto-mpg data, find the top 5 similar cars to toyota corona mark ii

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("auto-mpg (1).csv")

# Clean column names
df.columns = [c.strip().lower().replace(" ", "_") for c in df.columns]

# Replace '?' with NaN and convert columns to numeric
df.replace("?", np.nan, inplace=True)
for col in df.columns:
    if col != "car_name":
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Drop rows with missing values
df = df.dropna()

# Separate features and car names
X = df.drop(columns=["car_name"])
car_names = df["car_name"]

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Find index of Toyota Corona Mark II
target_index = car_names.str.lower().str.contains("toyota corona mark ii").idxmax()
target_vector = X_scaled[target_index]

# Compute Mean Squared Error (MSE)
mse = np.mean((X_scaled - target_vector) ** 2, axis=1)

# Create similarity results
results = pd.DataFrame({
    "car_name": car_names,
    "mse": mse
}).sort_values(by="mse")

# Top 5 similar cars (excluding itself)
top_5_similar = results.iloc[1:6]

print(top_5_similar)
```

### Question 2
Find completely unique/orthogonal car to toyota corona mark ii

```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Load dataset
df = pd.read_csv("auto-mpg (1).csv")

# Clean data
df.replace('?', np.nan, inplace=True)
df.dropna(inplace=True)

# Convert to numeric
numeric_cols = ['mpg', 'cylinders', 'displacement', 'horsepower',
                'weight', 'acceleration', 'model year', 'origin']
df[numeric_cols] = df[numeric_cols].astype(float)

# Feature matrix (NO standardization)
X = df[numeric_cols].values

# Target car
target_car = "toyota corona mark ii"
target_index = df[df['car name'] == target_car].index[0]

# Cosine similarity
cos_sim = cosine_similarity(X)

# Ignore self similarity
cos_sim[target_index, target_index] = 1

# Find most orthogonal car
most_orthogonal_index = np.argmin(cos_sim[target_index])

# Print the car name
print(df.iloc[most_orthogonal_index]['car name'])
```

### Question 3
Project the data to a 2-dimentional space and visualize using scatter plot

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# load your file
df = pd.read_csv("auto-mpg (1).csv")

# fix non-numeric horsepower
df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce")
df = df.dropna()

# separate names and features
names = df["car name"]
X = df.drop("car name", axis=1)

# normalize features
X_scaled = StandardScaler().fit_transform(X)

# project to 2D using PCA
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_scaled)

# scatter plot
plt.figure()
plt.scatter(X_2d[:, 0], X_2d[:, 1])
plt.xlabel("PC 1")
plt.ylabel("PC 2")
plt.title("2D Projection of Auto-MPG Data")
plt.show()
```

### Question 4
Write a function that returns dot product of two list of numbers

```python
def dot_product(a, b):
    return sum(x * y for x, y in zip(a, b))

result = dot_product([1, 2, 3], [4, 5, 6])
print(result)
```

### Question 5
Write a function that returns the projection of vector a onto vector b. Both a and b are given as list of numbers

```python
def projection(a, b):
    dot_ab = sum(x*y for x, y in zip(a, b))
    dot_bb = sum(x*x for x in b)
    return [(dot_ab / dot_bb) * x for x in b]

a = [1, 2, 3]
b = [4, 5, 6]

proj = projection(a, b)

print("Vector a:", a)
print("Vector b:", b)
print("Projection of a onto b:", proj)
```

### Question 6
Create a class Vector that supports addition, dot product, and magnitude.

```python
import math

class Vector:
    def __init__(self, values):
        self.values = values

    def add(self, other):
        return Vector([a + b for a, b in zip(self.values, other.values)])

    def dot(self, other):
        return sum(a * b for a, b in zip(self.values, other.values))

    def magnitude(self):
        return math.sqrt(sum(a * a for a in self.values))


# Example usage
v1 = Vector([1, 2, 3])
v2 = Vector([4, 5, 6])

print("v1 + v2 =", v1.add(v2).values)
print("Dot product =", v1.dot(v2))
print("Magnitude of v1 =", v1.magnitude())
```

### Question 7
Write a function that returns the L1 and L2 norms of a vector given as list

```python
import math

def vector_norms(v):
    l1 = sum(abs(x) for x in v)
    l2 = math.sqrt(sum(x*x for x in v))
    return l1, l2

v = [1, -2, 3]

result = vector_norms(v)

print("Vector:", v)
print("L1 norm:", result[0])
print("L2 norm:", result[1])
```
